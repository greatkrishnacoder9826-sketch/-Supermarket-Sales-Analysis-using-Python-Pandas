```python
Low = []
Medium = []
High = []

for i in df['Sales']:
    if i < 200:
        Low.append(i)
    elif i < 700:
        Medium.append(i)
    else:
        High.append(i)
```

```python
def categorize(sales):
    if sales < 200:
        return "Low"
    elif sales < 700:
        return "Medium"
    else:
        return "High"

df['Division'] = df['Sales'].apply(categorize)
df['Division']
```

```python
g = df[df['City'] == "Yangon"]
```

```python
df['Yangon_City_Categorization'] = g[g['Sales'] > g['Sales'].mean()]['Sales'].apply(lambda x: "Above Average" if x > g['Sales'].mean() else "Below Average")
df['Yangon_City_Categorization'].fillna("Unknown", inplace=True)
```

```python
g1 = df[df['City'] == "Naypyitaw"]
g1
```

```python
df['Naypyitaw_City_Categorization'] = g1[g1['Sales'] > g1['Sales'].mean()]['Sales'].apply(lambda x: "Above Average" if x > g1['Sales'].mean() else "Below Average")
df['Naypyitaw_City_Categorization'].fillna("Unknown", inplace=True)
```

```python
g2 = df[df['City'] == "Mandalay"]
g2
```

```python
df['Mandalay_City_Categorization'] = g2[g2['Sales'] > g2['Sales'].mean()]['Sales'].apply(lambda x: "Above Average" if x > g2['Sales'].mean() else "Below Average")
df['Mandalay_City_Categorization'].fillna("Unknown", inplace=True)
```

```python
df['City_Avg_Sales'] = df.groupby('City')['Sales'].transform('mean')

df['Sales_Label'] = np.where(df['Sales'] > df['City_Avg_Sales'], 
                              "Above Average", 
                              "Below Average")
```
